```python
def generate_create_table(df, table_name):
    sql_types = {
        "object": "VARCHAR(255)",
        "int64": "INT",
        "float64": "FLOAT",
        "datetime64[ns]": "DATETIME",
        "bool": "BOOLEAN"
    }
    
    columns = []
    for col_name, col_type in df.dtypes.items():
        sql_type = sql_types.get(str(col_type), "VARCHAR(255)")
        columns.append(f"{col_name} {sql_type}")
    
    columns_sql = ",\n    ".join(columns)
    create_table_sql = f"CREATE TABLE IF NOT EXISTS {table_name} (\n    {columns_sql}\n);"
    return create_table_sql

# Exemplo de uso
print(generate_create_table(df_customers, "customers"))
print(generate_create_table(df_geolocation, "geolocation"))
print(generate_create_table(df_order_items, "order_items"))
print(generate_create_table(df_order_payments, "order_payments"))
print(generate_create_table(df_order_reviews, "order_reviews"))
print(generate_create_table(df_orders, "orders"))
print(generate_create_table(df_products, "products"))
print(generate_create_table(df_sellers, "sellers"))
print(generate_create_table(df_product_category, "product"))
```
